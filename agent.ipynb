# Cover Letter Agent - calling `agent.py`

This notebook does not define the pipeline itself - it imports it from [`agent.py`](agent.py),
the standalone extraction of the pipeline in `cover_letter_v2.ipynb`. Any other notebook can
call the agent the same way: `from agent import load_inputs, run`.

`cover_letter_v2.ipynb` is untouched and keeps its own copy of the same pipeline for
interactive editing - the two are not linked, so a prompt/node change made in one won't show up
in the other unless copied over by hand.

```python
from IPython.display import Markdown, display

from agent import load_inputs, run

inputs = load_inputs()
```

`run()` tries the `langgraph dev` server first (so the run shows up as a thread in Studio),
falling back to an in-process `agent.graph.invoke(...)` if the server isn't running.

```python
result = run(inputs)
display(Markdown(result["final_letter"]))
```

## Inspect the pieces

```python
print("=" * 70); print("RESEARCH NOTES AND CANDIDATE MOTIVATIONS")
print(result["research_notes"])

print("\n" + "=" * 70); print("QUALIFICATIONS -> VALUE TO THE TEAM")
for item in result["qualifications"]:
    print(f"\n  [{item['relevance']}/5] {item['qualification']}")
    print(f"       evidence : {item['evidence'][:110]}")
    print(f"       value    : {item['value_to_team'][:110]}")

print("\n" + "=" * 70); print("PARAGRAPHS AS WRITTEN")
for label, key in [("INTRO", "para_intro"), ("BODY", "para_body"), ("CLOSE", "para_close")]:
    print(f"\n--- {label} ({len(result[key].split())} words)\n{result[key]}")

if result.get("changes"):
    print("\n" + "=" * 70); print("CHANGES MADE BY ASSEMBLE")
    for change in result["changes"]:
        print("  - " + change)
```

```python
print(f"SOURCES ({len(result['sources'])})")
for source in result["sources"][:12]:
    print(f"  - {source['title'][:65]}\n    {source['url']}")
```
